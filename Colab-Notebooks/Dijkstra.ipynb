### **¿Qué es una heurística y qué significa que sea admisible?**

#### La heurística es una estrategia de resolución de problemas que se enfoca en la búsqueda rápida con menor exactitud [1], tiene como propósito guiar el algoritmo a lo largo de líneas que tengan alta probabilidad de éxito [2]. En el ámbito de la inteligencia artifical se considera como un atajo para la resolución más rápida y eficiente.
#### Se considera admisible cuando el costo predicho es menor al costo real, es decir, si el costo predicho para llegar al objetivo es sobreestimado (mayor al real), no se considera como admisible [3].

### **Distancia Manhattan**

#### Mide la distancia entre dos puntos sumando las diferencias absolutas en cada dimensión [4], se calcula como:

$$
d=\sum_{i=1}^{n}|x_i-y_i|
$$

#### donde:

$$ \begin{aligned} n &: \text{número de dimensiones} \\ x_i &: \text{coordenadas del primer punto} \\ y_i &: \text{coordenadas del segundo punto} \end{aligned} $$

### **Distancia euclidiana**

#### Equivalente a la línea recta entre dos puntos [4], se calcula como:

$$
d=\sqrt{\sum_{i=1}^{n}(x_i-y_i)^2}
$$

#### donde:

$$ \begin{aligned} n &: \text{número de dimensiones} \\ x_i &: \text{coordenadas del primer punto} \\ y_i &: \text{coordenadas del segundo punto} \end{aligned} $$

### **Distancia en línea recta**

####La distancia en línea recta entre dos puntos del plano cartesiano corresponde a la longitud del segmento que une ambos puntos y se calcula a partir de sus coordenadas, aplicando el teorema de Pitágoras [5].

$$
d=\sqrt{(x_2-x_1)^2 + (y_2-y_1)^2}
$$



### **Discrepancia posicional**

#### Se refiere a la diferencia existente entre la posición observada o registrada de un objeto o punto y su posición de referencia. Esta diferencia puede determinarse mediante la comparación de sus coordenadas espaciales, permitiendo identificar las variaciones entre ambas ubicaciones [6].



#### Fórmula de discrepancia posicional
$$
D_p=\sqrt{(x_o-x_r)^2 + (y_o-y_r)^2}
$$

#### donde:

$$ \begin{aligned} D_p &: \text{magnitud de la discrepancia posicional} \\ x_o, y_o &: \text{coordenadas de la posición observada o registrada} \\ x_r, y_r &: \text{coordenadas de la posición de referencia} \end{aligned} $$

### **Dijkstra**

```python
def dijkstra(graph, start):
  # get the size of the graph and initialize distances as inf
  num_nodes = len(graph)
  distances = [float('inf')] * num_nodes

  # initialize start node with distance 0 and frontier with start node
  distances[start] = 0
  frontier = [(start, 0)]

  # track visited nodes
  visited = [False] * num_nodes

  # continue until frontier is empty
  while frontier:
    min_dist = 0

    # search for the lowest cost (manual priority queue)
    for i in range(1, len(frontier)):
      if frontier[i][1] < frontier[min_dist][1]:
        min_dist = i

    # remove the node with the minimum cost
    node, cost = frontier.pop(min_dist)

    # skip if already visited
    if visited[node]:
      continue

    # else mark it as visited and store shortest distance
    visited[node] = True
    distances[node] = cost

    # explore all the neighbors of the current node
    for neighbor in range(num_nodes):
      if graph[node][neighbor] is not None and neighbor != start:
        new_cost = distances[node] + graph[node][neighbor]

        # update if a shorter path is found
        if new_cost < distances[neighbor]:
            distances[neighbor] = new_cost

            # add the neighbor to the frontier
            frontier.append((neighbor, new_cost))

  # return shortest distances
  return distances
```

```python
graph = [
    [None, 4,    2,    None],
    [None, None, None, 5   ],
    [None, 1,    None, 8   ],
    [None, None, None, None]
]
result = dijkstra(graph, 0)

print("\nDistances:")
for i, d in enumerate(result):
    print(f"0 -> {i} = {d}")
```

### **Dijkstra with heuristics**

```python
def no_heuristic(node, end):
  return 0
```

```python
def manhattan_distance(node, end):
  return abs(node[0] - end[0]) + abs(node[1] - end[1])
```

```python
import math
def euclidean_distance(node, end):
  return math.sqrt((node[0] - end[0])**2+(node[1] - end[1])**2)
```

```python
HEURISTIC_FUNCTIONS = {
  "Dijkstra ": no_heuristic,
  "Manhattan": manhattan_distance,
  "Euclidean": euclidean_distance,
}
```

```python
def dijkstra(graph, start, end, heuristic, positions):
  # get the size of the graph and initialize distances as inf
  num_nodes = len(graph)
  distances = [float('inf')] * num_nodes

  # heuristic function selected
  h = HEURISTIC_FUNCTIONS[heuristic]

  # initialize start node with distance 0 and frontier with start node
  # frontier stores (node, priority) where priority = g + h
  distances[start] = 0
  frontier = [(start, 0)]

  # track visited nodes, visit order and previous node
  visited = [False] * num_nodes
  visited_order = []
  previous = [None] * num_nodes

  # continue until frontier is empty
  while frontier:
    min_dist = 0

    # search for the lowest cost (manual priority queue)
    for i in range(1, len(frontier)):
      if frontier[i][1] < frontier[min_dist][1]:
        min_dist = i

    # remove the node with the minimum cost
    node, cost = frontier.pop(min_dist)

    # skip if already visited
    if visited[node]:
      continue

    # mark it as visited (distances[node] already has g; cost is g + h)
    visited[node] = True
    visited_order.append(node)

    # stop when the end is reached
    if node == end:
      break

    # explore all the neighbors of the current node
    for neighbor in range(num_nodes):
      if graph[node][neighbor] is not None and not visited[neighbor]:
        new_cost = distances[node] + graph[node][neighbor]

        # update if a shorter path is found
        if new_cost < distances[neighbor]:
          distances[neighbor] = new_cost
          previous[neighbor] = node

          # add the neighbor to the frontier with priority g + h
          priority = new_cost + h(positions[neighbor], positions[end])
          frontier.append((neighbor, priority))

  # no path: the end was never reached
  if not visited[end]:
    return distances, visited_order, []

  # rebuild the path going back from the end to the start
  path = [end]
  while path[-1] != start:
    path.append(previous[path[-1]])
  path.reverse()

  return distances, visited_order, path
```

```python
graph = [
    [None, 4,    2,    None],
    [None, None, None, 5   ],
    [None, 1,    None, 8   ],
    [None, None, None, None]
]

# (x, y) of each node
positions = [(0, 0), (2, 1), (2, 0), (2, 6)]

for heuristic in HEURISTIC_FUNCTIONS:
  distances, visited_order, path = dijkstra(graph, 0, 3, heuristic, positions)
  print(heuristic, "path:", path, "| cost:", distances[3], "| visited:", visited_order)
```
